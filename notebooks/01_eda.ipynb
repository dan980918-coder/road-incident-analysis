```python
import sys, pathlib
sys.path.insert(0, str(pathlib.Path.cwd().parent))
import pandas as pd
from src.load_data import load
from src.plot_config import set_korean_style
import matplotlib.pyplot as plt

set_korean_style()
pd.set_option('display.max_columns', 30)
```

## 데이터 로드

```python
df = load()
print('shape:', df.shape)
df.head()
```

## 데이터 타입 및 결측치

```python
info = pd.DataFrame({
    'dtype': df.dtypes,
    'null_count': df.isna().sum(),
    'null_pct': (df.isna().mean() * 100).round(2)
})
info
```

## 기간 범위 확인

```python
print('시작:', df['돌발일시'].min())
print('종료:', df['돌발일시'].max())
```

## 돌발구분 / 돌발상세구분 분포

```python
display(df['돌발구분'].value_counts())
display(df['돌발상세구분'].value_counts())
```
